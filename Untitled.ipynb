```python
# importing eda libries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
# loading the datasets
df = pd.read_csv("Social_media_impact_on_life.csv")

df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
# checking wheather there are any null values
df.isnull().sum()
```

```python
df['Perceived_Stress_Score'].head()
```

```python
df["Overall_Impact"].value_counts()
```

```python
df["Overall_Impact"].unique()
```

```python
df["Impact_Binary"] = df["Overall_Impact"].apply(
    lambda x: 1 if x == "Beneficial" else 0
)
```

```python
df["Impact_Binary"].value_counts()
```

```python
# this will show the side bye side comparsion of two features
pd.crosstab(df["Overall_Impact"], df["Impact_Binary"])
```

```python
df["Impact_Binary"].value_counts()
```

```python
y = df["Impact_Binary"]
```

```python
X = df.drop(columns=["Overall_Impact", "Impact_Binary"])
```

```python
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
X.head()
```

```python
X.dtypes
```

```python
#Student_ID is just an identifier. It doesn't provide useful information for predicting the impact.
```

```python
X.drop('Student_ID', axis = 1, inplace = True)
```

```python
for col in X.select_dtypes(include="object").columns:
    print("\n", col)
    print(X[col].unique())
```

```python
categorical_cols = X.select_dtypes(include="object").columns

print(categorical_cols)
```

```python
X = pd.get_dummies(
    X,
    columns=categorical_cols, dtype = int,
    drop_first=True
)
```

```python
X.head()
```

```python
bool_cols = X.select_dtypes(include="bool").columns

X[bool_cols] = X[bool_cols].astype(int)
```

```python
import numpy as np
np.isinf(X.select_dtypes(include=np.number)).sum()
```

```python
X.isnull().sum()
```

```python
# filling the nah values using the mode
X["Perceived_Stress_Score"] = X["Perceived_Stress_Score"].fillna(
    X["Perceived_Stress_Score"].median()
)

X["Academic_Performance_GPA"] = X["Academic_Performance_GPA"].fillna(
    X["Academic_Performance_GPA"].median()
)
```

```python
X = X.astype(int)
```

```python
X.info()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
X_train.head()
```

```python
y_train.head()
```

```python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)

print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
print("Training target:")
print(y_train.value_counts())

print("\nTesting target:")
print(y_test.value_counts())
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)

model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
print(y_pred[:20])
```

```python
for i in y_pred[:20]:
    if i == 1:
        print("Beneficial")
    else:
        print("Neutral/Negative")
```

```python
# checking the model_acuuracy
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
```

```python
from sklearn.metrics import precision_score, recall_score

precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)

print("Precision:", precision)
print("Recall:", recall)
```

```python
from sklearn.metrics import f1_score

f1 = f1_score(y_test, y_pred)

print("F1 Score:", f1)
```

```python
df_lg = (print("Accuracy :", accuracy),
print("Precision:", precision),
print("Recall   :", recall),
print("F1 Score :", f1))
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
#Next basic step: classification report
#Instead of calculating each metric separately, we can get them together:
from sklearn.metrics import classification_report

print(classification_report(
    y_test,
    y_pred,
    target_names=["Neutral/Negative", "Beneficial"]
))
```

```python
# in next step we are using a better model like decision tree
from sklearn.tree import DecisionTreeClassifier

dt_model = DecisionTreeClassifier(random_state=42)

dt_model.fit(X_train, y_train)
```

```python
dt_pred = dt_model.predict(X_test)
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score, accuracy_score
dt_accuracy = accuracy_score(y_test,dt_pred)
dt_precision = precision_score(y_test, dt_pred)
dt_recall = recall_score(y_test, dt_pred)
dt_f1 = f1_score(y_test, dt_pred)
df_dt = (
print("Accuracy :", dt_accuracy),
print("Precision:", dt_precision),
print("Recall   :", dt_recall),
print("F1 Score :", dt_f1))
```

```python
# we were using a diffrent model the random forest classsifier
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

rf_model.fit(X_train, y_train)
```

```python
rf_pred = rf_model.predict(X_test)
```

```python
rf_accuracy = accuracy_score(y_test, rf_pred)
rf_precision = precision_score(y_test, rf_pred)
rf_recall = recall_score(y_test, rf_pred)
rf_f1 = f1_score(y_test, rf_pred)

print("Accuracy :", rf_accuracy)
print("Precision:", rf_precision)
print("Recall   :", rf_recall)
print("F1 Score :", rf_f1)
```

```python
# we were also using an diffrent model which is knn 
from sklearn.neighbors import KNeighborsClassifier

knn_model = KNeighborsClassifier(n_neighbors=5)

knn_model.fit(X_train, y_train)
```

```python
knn_pred = knn_model.predict(X_test)
```

```python
knn_accuracy = accuracy_score(y_test, knn_pred)
knn_precision = precision_score(y_test, knn_pred)
knn_recall = recall_score(y_test, knn_pred)
knn_f1 = f1_score(y_test, knn_pred)

print("Accuracy :", knn_accuracy)
print("Precision:", knn_precision)
print("Recall   :", knn_recall)
print("F1 Score :", knn_f1)
```
